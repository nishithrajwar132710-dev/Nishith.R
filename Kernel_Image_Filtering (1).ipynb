# Kernel Based Image Filtering
**Aim:** To implement image filtering using different kernels (blur, sharpen, edge detection, emboss).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

### Load image

```python
# sample image, to use own image write: img = cv2.imread('photo.jpg')
img = data.astronaut()
img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')
plt.show()
```

### Define kernels

```python
identity = np.array([[0,0,0],
                     [0,1,0],
                     [0,0,0]])

box_blur = np.ones((5,5), np.float32) / 25

gaussian = np.array([[1,2,1],
                     [2,4,2],
                     [1,2,1]]) / 16

sharpen = np.array([[0,-1,0],
                    [-1,5,-1],
                    [0,-1,0]])

edge = np.array([[-1,-1,-1],
                 [-1,8,-1],
                 [-1,-1,-1]])

sobel_x = np.array([[-1,0,1],
                    [-2,0,2],
                    [-1,0,1]])

sobel_y = np.array([[-1,-2,-1],
                    [0,0,0],
                    [1,2,1]])

emboss = np.array([[-2,-1,0],
                   [-1,1,1],
                   [0,1,2]])

kernels = {"Identity": identity, "Box Blur": box_blur, "Gaussian Blur": gaussian,
           "Sharpen": sharpen, "Edge Detection": edge, "Sobel X": sobel_x,
           "Sobel Y": sobel_y, "Emboss": emboss}
```

### Apply kernels

```python
plt.figure(figsize=(12,8))

i = 1
for name, k in kernels.items():
    out = cv2.filter2D(img, -1, k.astype(np.float32))
    plt.subplot(2,4,i)
    plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
    plt.title(name)
    plt.axis('off')
    i += 1

plt.tight_layout()
plt.show()
```

**Conclusion:** Different kernels give different effects on the same image. Blur kernels smooth the image, sharpen kernel makes edges clearer, and edge and Sobel kernels detect edges.
